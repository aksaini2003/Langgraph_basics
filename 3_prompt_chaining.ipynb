```python
#here we are trying to make a blog writing with the help of the promptchaining in the langgraph 

# structure is as follows 

#  topic ---> create_outline(llm_call1) ---> write_blog(llm_call2) ---> return blog 
```

```python
from langchain_groq import ChatGroq 
from langgraph.graph import StateGraph ,START,END 
from typing import TypedDict 
from dotenv import load_dotenv
```

```python
class Blog_state(TypedDict):
    topic: str 
    outline: str 
    blog: str 
```

```python
graph=StateGraph(Blog_state) #making a graph 
```

```python
#lets make function for the llm call 
def create_outline(state: Blog_state)-> Blog_state:
    topic=state['topic']
    prompt=f'''You have to create an outline for the following topic of blog 
    
    
    blog topic -- {topic}'''
    #call llm 
    load_dotenv() 
    llm=ChatGroq(model='llama-3.3-70b-versatile')
    response=llm.invoke(prompt).content 
    state['outline']=response 
    return state 


#lets make a another funciton for writing the actual blog 
def write_blog(state: Blog_state)-> Blog_state:
    outline=state['outline']
    #now we have to write the blog on the given topic's outline 
    prompt=f'''You are a blog writer and you have to write a blog according to the following outline
    
    outline -- {outline}'''
    load_dotenv() 
    llm=ChatGroq(model='Llama3-70b-8192')
    response=llm.invoke(prompt).content
    state['blog']=response 
    return state
```

```python
#we have successfully implemented the code for the node funciton 
#lets make a graph for all of the work 
graph.add_node('outline_writer',create_outline)
graph.add_node('blog_writer',write_blog)


#define edges 
graph.add_edge(START,'outline_writer')
graph.add_edge('outline_writer','blog_writer')
graph.add_edge('blog_writer',END)
```

```python
#lets make a workflow for this graph 
blog_writing_workflow=graph.compile()
```

```python
#lets make a initial state and then invoke our workflow 
initial_state={'topic':'Global Warming'}

final_state=blog_writing_workflow.invoke(initial_state)
print(final_state) 
```

```python
#lets print the blog and its outline which is created by our llm model 
from IPython.display import display,Markdown 
display(Markdown(final_state['outline']))

display(Markdown(final_state['blog']))


#how we can improve it we can improve it by providing the websearch to our model 
```
