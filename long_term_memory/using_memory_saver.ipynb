```python
from langgraph.store.memory import InMemoryStore

from langchain_groq import ChatGroq 
from langgraph.graph import StateGraph,START,END 
from langchain_google_genai import GoogleGenerativeAIEmbeddings

from dotenv import load_dotenv 
```

```python
load_dotenv()
```

```python
storage=InMemoryStore()

# #here we see 4 methods 
# 1. storage .put()
# 2. retrieving  .get()
# 3. searching .search() 
# 4. deletion .delete()
```

## Storage

```python
#here we will use the above storage object for storing the name spaces 
#namespace is just a directory inside in this storage 
name_space1=('user','u1') #here  user is the parent folder and u1 is another folder in it 
name_space2=('user','u2') #just above but another folder 

```

```python
#lets add data or memories in this store or storage 
storage.put(name_space1,"1",{'data':'Aashish likes to play  Cricket every sunday'})
storage.put(name_space1,'2',{'data':'Aashish likes bourbon biscuits'})

#here the key is use to identify the each data or entry uniquely in the store 
```

```python
#now lets also add the entries for the name_space2 
storage.put(name_space2,"1",{'data':'Ram likes to play  Badminton every sunday'})
storage.put(name_space2,'2',{'data':'Ram likes Dark Fantasy biscuits'})
```

## Retrieval


```python
# 1.TO GET THE SPECIFIC ENTRY FROM THE NAMESPACE USING THE KEY 
storage.get(namespace=name_space1,key='1') #we have to provide the key for getting the specific entry in it 
```

```python
#2. TO RETRIEVE ALL THE MEMORIES OF A NAMESPACE 
items =storage.search(name_space1)
```

```python
for data in items:
    print(data)
```

```python
for data in items:
    print(data.value['data'])
```

## Deletion 

```python
#lets delete the 1 key data in the name_space2 
```

```python
storage.delete(namespace=name_space2,key='2') 
```

```python
#lets print the data after the deletion 
storage.search(name_space2)
#successfully deleted the data inside the namespace with 2 as a key
```

## Semantic Search 

<html> As we have see how to get the particular data with a key or all the data of the namespaces.
<br>But what if we have to search on the basis of the meaning of the query, then we have to use 
the Semantic search approach.  </html>

```python
#we need a embedding model for this work 
embedding_model=GoogleGenerativeAIEmbeddings(model='models/text-embedding-004') 
```

```python
embedding=embedding_model.embed_query('hello my name is aashisi') #working 
```

```python
len(embedding)
```

```python
#now lets see how to use the store for this work 

#we have to pass an embedding model in it. 
sem_store=InMemoryStore(index={'embed':embedding_model,'dims':768})
```

```python
namespace=('user','aashish')
```

```python
#now lets store a particular set of lines using the put()
sem_store.put(namespace, "2", {"data": "User likes examples in Python"})
sem_store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
sem_store.put(namespace, "3", {"data": "User usually works late at night"})
sem_store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
sem_store.put(namespace, "5", {"data": "User is learning machine learning"})
sem_store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
sem_store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
sem_store.put(namespace, "8", {"data": "User is based in India"})
sem_store.put(namespace, "9", {"data": "User likes real-world analogies"})
sem_store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
#now lets do semantic searching 
results=sem_store.search(namespace,query='when the user usually works',limit=1) 

```

```python
for item in results:
    print(item)
    print(item.value['data'])  #yeah it is working fine now lets try again 
```

```python
res=sem_store.search(namespace,query='which types of answers user prefers',limit=2) 
```

```python
for item in res:
    print(item.value)
```
