```python
#here we are going to create the first langgraph application 
#here we make a bmi calculator that take height and weight as input 
from langgraph.graph import StateGraph,START,END 
from typing import TypedDict 
```

```python
#first define the state 
class BMIState(TypedDict):
    height: float 
    weight: float 
    bmi: float 
    category: str 
    
```

```python
#function for the node 
def calculate_bmi(state: BMIState)-> BMIState: 
    height=state['height']
    weight=state['weight']
    bmi=weight/(height*height)
    #lets update the state 
    state['bmi']=round(bmi,2)
    return state
    
```

```python
# lets add a new ability to show the body category according to the user's bmi  
# Major adult BMI classifications are underweight (under 18.5 kg/m 2), normal weight (18.5 to 24.9), overweight (25 to 29.9), and obese (30 or more).
def body_category(state: BMIState)-> BMIState:
    if state['bmi']<18.5:
        state['category']='Under Weight' 
    elif state['bmi']>=18.5 and state['bmi']<=24.9:
        state['category']='Normal Weight' 
    elif state['bmi']>=25 and state['bmi']<29.9:
        state['category']='Overweight' 
    else:
        state['category']='Obese' 
    return state 
```

```python
#lets define the graph 
graph=StateGraph(BMIState)
#here we have to pass the State 

#lets add nodes in it 
graph.add_node('calculate_BMI',calculate_bmi) #2 argument 1st - node name , argument 2nd - acutal function 

graph.add_node('body_category',body_category)
#lets add the edges 
graph.add_edge(START,'calculate_BMI')
graph.add_edge('calculate_BMI','body_category')

graph.add_edge('body_category',END)


```

```python
#lets compile it  and make it instance 

workflow=graph.compile()
#so it is ready for execution 
#lets see our graph 
workflow
```

```python
#so it is ready for execution 
initial_state={'weight':58,'height':1.73}
final_state=workflow.invoke(initial_state)
print(final_state)
#so our workflow is working fine 
```
